```python
import pandas as pd 
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("latency.csv")
df.head()
```

```python
p99 = np.percentile(df['latency'], 99)

sns.ecdfplot(data=df, x='latency')

plt.axvline(p99, color='black', linestyle='--', alpha=0.5)
plt.axhline(0.99, color='black', linestyle=':', alpha=0.5)
plt.scatter(p99, 0.99, color='red', zorder=5)

plt.title("CDF plot e2e latency of requests")
plt.show()
```
